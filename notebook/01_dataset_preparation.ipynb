# Experiment Record-dataset download

## Download RePAIR dataset
- source: https://repairproject.github.io/RePAIR_dataset/
- download tool: aria2c
  - support multiple threads download
  - continue to download if failed 
  - setup on mac:
    - using Homebrew: `brew install aria2`
    - installed check: `aria2c --version`
    - parameters:
      - `-x`: max connection per server
      - `-s`: split file to segment or chunk
      - real concurrent connection: `min(x, s)`
- 2D dataset
  - files:   
    - ground truth:
      - command: `find . -maxdepth 1 -type f -iname "*.txt" | wc -l`           
      - .txt annotation: 100 items  
    - fragment images:
      - command: `find . -maxdepth 1 -type d -mindepth 1 | wc -l` 
      - assembled_objects: 100 items
      - isolated_pieces: 54 items
  - .zip file size: `1.1GB`
  - download command: `aria2c -x 16 -s 16 -k 5M "https://zenodo.org/records/13993089/files/2D_Fragments.zip?download=1"`
- 3D dataset
  - files: not download 
  - .zip file size: `53.5GB`
  - download command: `aria2c -x 16 -s 16 -k 5M "https://zenodo.org/records/13993089/files/3D_Fragments.zip?download=1"`
- storage path: raw_data (not upload)
